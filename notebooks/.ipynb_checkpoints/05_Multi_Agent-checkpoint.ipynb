```python
import requests
import pandas as pd

from langgraph.graph import StateGraph, END
from typing import TypedDict
```

```python
df = pd.read_csv("../data/sales_data.csv")

print(df.head())

print()

print("Rows:", len(df))
```

```python
def ask_qwen(prompt):

    response = requests.post(
        "http://localhost:11434/api/generate",
        json={
            "model": "qwen3:4b",
            "prompt": prompt,
            "stream": False
        }
    )

    return response.json()["response"]

print("Qwen Ready ✅")
```

```python
print(
    ask_qwen(
        "Say Hello in one sentence."
    )
)
```

```python
class AgentState(TypedDict):
    question: str
    answer: str
```

```python
def sales_agent(state):

    prompt = f"""
You are a Senior Sales Analyst.

Answer this question professionally:

{state["question"]}
"""

    answer = ask_qwen(prompt)

    return {
        "question": state["question"],
        "answer": answer
    }

print("Sales Agent Ready ✅")
```

```python
def marketing_agent(state):

    prompt = f"""
You are a Senior Marketing Consultant.

Answer this question professionally:

{state["question"]}
"""

    answer = ask_qwen(prompt)

    return {
        "question": state["question"],
        "answer": answer
    }

print("Marketing Agent Ready ✅")
```

```python
def finance_agent(state):

    prompt = f"""
You are a Senior Finance Consultant.

Answer this question professionally:

{state["question"]}
"""

    answer = ask_qwen(prompt)

    return {
        "question": state["question"],
        "answer": answer
    }

print("Finance Agent Ready ✅")
```

```python
result = sales_agent(
    {
        "question": "How can I increase sales?"
    }
)

print(result["answer"])
```

```python
workflow = StateGraph(AgentState)

workflow.add_node("sales", sales_agent)

workflow.set_entry_point("sales")

workflow.add_edge("sales", END)

graph = workflow.compile()

print("Workflow Created ✅")
```

```python
result = graph.invoke(
    {
        "question": "How can Digital Whopper increase sales?"
    }
)

print(result["answer"])
```

```python
result = graph.invoke(
    {
        "question": "What are good ways to improve customer retention?"
    }
)

print(result["answer"])
```

```python
def supervisor(state):

    question = state["question"].lower()

    if any(word in question for word in [
        "sales",
        "revenue",
        "customer",
        "product",
        "profit"
    ]):
        return {"next": "sales"}

    elif any(word in question for word in [
        "marketing",
        "seo",
        "google ads",
        "campaign",
        "lead"
    ]):
        return {"next": "marketing"}

    elif any(word in question for word in [
        "finance",
        "cost",
        "expense",
        "budget"
    ]):
        return {"next": "finance"}

    else:
        return {"next": "sales"}

print("Supervisor Ready ✅")
```

```python
print(
    supervisor(
        {
            "question":"How can I improve SEO?"
        }
    )
)

print(
    supervisor(
        {
            "question":"Revenue is falling."
        }
    )
)

print(
    supervisor(
        {
            "question":"Reduce company expenses."
        }
    )
)
```

```python
workflow = StateGraph(AgentState)

workflow.add_node("sales", sales_agent)
workflow.add_node("marketing", marketing_agent)
workflow.add_node("finance", finance_agent)

workflow.set_entry_point("sales")

workflow.add_edge("sales", END)
workflow.add_edge("marketing", END)
workflow.add_edge("finance", END)

graph = workflow.compile()

print("Enterprise Workflow Ready ✅")
```

```python
from typing import TypedDict

class AgentState(TypedDict):
    question: str
    answer: str
    next: str
```

```python
def router(state):
    return state["next"]
```

```python
workflow = StateGraph(AgentState)

workflow.add_node("supervisor", supervisor)
workflow.add_node("sales", sales_agent)
workflow.add_node("marketing", marketing_agent)
workflow.add_node("finance", finance_agent)

workflow.set_entry_point("supervisor")

workflow.add_conditional_edges(
    "supervisor",
    router,
    {
        "sales": "sales",
        "marketing": "marketing",
        "finance": "finance"
    }
)

workflow.add_edge("sales", END)
workflow.add_edge("marketing", END)
workflow.add_edge("finance", END)

graph = workflow.compile()

print("Real Multi-Agent Graph Ready ✅")
```

```python
result = graph.invoke(
    {
        "question":"Revenue is decreasing."
    }
)

print(result["answer"])
```

```python
result = graph.invoke(
    {
        "question":"How can SEO improve?"
    }
)

print(result["answer"])
```

```python
result = graph.invoke(
    {
        "question":"How can we reduce expenses?"
    }
)

print(result["answer"])
```
