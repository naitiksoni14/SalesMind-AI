```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/sales_data.csv")

df.head()
```

```python
print("Shape of Dataset:", df.shape)

print("\nColumns:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)
```

```python
total_revenue = df["Revenue"].sum()

print(f"Total Revenue: ₹{total_revenue:,}")
```

```python
average_revenue = df["Revenue"].mean()

print(f"Average Revenue per Sale: ₹{average_revenue:,.2f}")
```

```python
highest_revenue = df["Revenue"].max()

print(f"Highest Revenue: ₹{highest_revenue:,}")
```

```python
lowest_revenue = df["Revenue"].min()

print(f"Lowest Revenue: ₹{lowest_revenue:,}")
```

```python
total_sales = len(df)

print(f"Total Sales: {total_sales}")
```

```python
service_revenue = (
    df.groupby("Service")["Revenue"]
      .sum()
      .sort_values(ascending=False)
)

service_revenue
```

```python
service_revenue.plot(
    kind="bar",
    figsize=(10,5)
)

plt.title("Revenue by Service")
plt.xlabel("Service")
plt.ylabel("Revenue")

plt.xticks(rotation=45)

plt.show()
```

```python
# ==========================================
# CITY ANALYSIS
# ==========================================
```

```python
city_revenue = (
    df.groupby("City")["Revenue"]
      .sum()
      .sort_values(ascending=False)
)

city_revenue
```

```python
city_revenue.plot(
    kind="bar",
    figsize=(10,5)
)

plt.title("Revenue by City")
plt.xlabel("City")
plt.ylabel("Revenue")

plt.xticks(rotation=45)

plt.show()
```

```python
# ==========================================
# TOP CUSTOMERS
# ==========================================
```

```python
top_customers = (
    df.groupby("Company_Name")["Revenue"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

top_customers
```

```python
top_customers.plot(
    kind="bar",
    figsize=(12,5)
)

plt.title("Top 10 Customers")
plt.xlabel("Company")

plt.ylabel("Revenue")

plt.xticks(rotation=90)

plt.show()
```
