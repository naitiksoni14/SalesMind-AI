```python
import pandas as pd
import numpy as np
from faker import Faker
import random
from datetime import datetime, timedelta
```

```python
# Initialize Faker
fake = Faker("en_IN")

# Make our random data reproducible
random.seed(42)
np.random.seed(42)

print("SalesMind AI Environment Ready ✅")
```

```python
services = [
    "SEO",
    "Google Ads",
    "Meta Ads",
    "Website Development",
    "AI Chatbot Development",
    "Social Media Marketing",
    "Content Marketing",
    "Email Marketing"
]

print(services)
```

```python
order_id = 1001

print(order_id)
```

```python
customer_name = fake.name()

print(customer_name)
```

```python
company_name = fake.company()

print(company_name)
```

```python
sales_record = {
    "Order_ID": order_id,
    "Customer_Name": customer_name,
    "Company_Name": company_name
}

print(sales_record)
```

```python
sales_record = {
    "Order_ID": order_id,
    "Customer_Name": customer_name,
    "Company_Name": company_name,
    "Service": random.choice(services),
    "City": fake.city(),
    "Revenue": random.randint(20000, 300000)
}

print(sales_record)
```

```python
def generate_sales_record(order_id):

    services = [
        "SEO",
        "Google Ads",
        "Meta Ads",
        "Website Development",
        "AI Chatbot Development",
        "Social Media Marketing",
        "Content Marketing",
        "Email Marketing"
    ]

    cities = [
        "Jaipur",
        "Delhi",
        "Mumbai",
        "Bangalore",
        "Hyderabad",
        "Pune",
        "Ahmedabad",
        "Kolkata"
    ]

    return {

        "Order_ID": order_id,

        "Customer_Name": fake.name(),

        "Company_Name": fake.company(),

        "Service": random.choice(services),

        "City": random.choice(cities),

        "Revenue": random.randint(20000, 300000)

    }
```

```python
record = generate_sales_record(1001)

record
```

```python
sales_data = []

for order_id in range(1001, 11001):
    sales_data.append(generate_sales_record(order_id))

print("Total Records Generated:", len(sales_data))
```

```python
df = pd.DataFrame(sales_data)

df.head()
```

```python
df.to_csv("../data/sales_data.csv", index=False)

print("Dataset Saved Successfully ✅")
```
