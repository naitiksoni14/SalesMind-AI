```python
import pandas as pd
import requests
import json
```

```python
df = pd.read_csv("../data/sales_data.csv")

df.head()
```

```python
total_revenue = df["Revenue"].sum()
average_revenue = df["Revenue"].mean()
highest_revenue = df["Revenue"].max()
lowest_revenue = df["Revenue"].min()
total_sales = len(df)

best_service = (
    df.groupby("Service")["Revenue"]
    .sum()
    .idxmax()
)

best_city = (
    df.groupby("City")["Revenue"]
    .sum()
    .idxmax()
)

print("KPIs Ready ✅")
```

```python
business_summary = f"""
Business Summary

Total Revenue: ₹{total_revenue:,.0f}

Average Revenue: ₹{average_revenue:,.2f}

Highest Revenue: ₹{highest_revenue:,.0f}

Lowest Revenue: ₹{lowest_revenue:,.0f}

Total Sales: {total_sales}

Best Service:
{best_service}

Best City:
{best_city}

You are a Senior Business Consultant.

Based on the above KPIs provide:

1. Executive Summary

2. Business Insights

3. Business Risks

4. Recommendations

Write professionally.
"""

print(business_summary)
```

```python
url = "http://localhost:11434/api/generate"

payload = {
    "model": "qwen3:4b",
    "prompt": business_summary,
    "stream": False
}

response = requests.post(url, json=payload)

print(response.json()["response"])
```
